# Open AI Video Studio — Hardware Check
Run this notebook interactively in Colab before downloading model weights.


```python
import platform, subprocess, torch, psutil
print('=' * 48)
print('OPEN AI VIDEO STUDIO - SYSTEM CHECK')
print('=' * 48)
print('Python:', platform.python_version())
print('PyTorch:', torch.__version__)
if torch.cuda.is_available():
    print('\nGPU AVAILABLE')
    for i in range(torch.cuda.device_count()):
        gpu = torch.cuda.get_device_properties(i)
        print(f'GPU {i}: {gpu.name}')
        print(f'VRAM: {gpu.total_memory / 1024**3:.2f} GB')
    print('CUDA:', torch.version.cuda)
else:
    print('\nNO GPU DETECTED — change Runtime type to GPU if available.')
ram = psutil.virtual_memory()
print(f'RAM: {ram.total / 1024**3:.2f} GB total; {ram.available / 1024**3:.2f} GB available')
print('\nDisk:')
subprocess.run(['df', '-h', '/content'])
print('\nSYSTEM CHECK COMPLETE')
```
